```python
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
from IPython.display import clear_output

# Configuración de estilo
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
class SistemaControlPanIntegral:
    def __init__(self):
        # Rangos definidos en nuestro análisis técnico
        self.limite_temp = 26.0  # Celsius
        self.humedad_min = 75.0  # % HR
        self.humedad_max = 85.0  # % HR
        self.torque_objetivo = 45.0 # Nm (Punto de desarrollo de gluten)

    def simular_lectura(self):
        """Simula datos de sensores con ruido aleatorio"""
        return {
            "Temperatura": np.round(np.random.normal(25, 1.5), 2),
            "Humedad": np.round(np.random.uniform(70, 90), 2),
            "Torque": np.round(np.random.uniform(30, 50), 2)
        }

    def verificar_alertas(self, datos):
        alertas = []
        if datos["Temperatura"] > self.limite_temp:
            alertas.append("⚠️ ALERTA: Temperatura elevada. Activando refrigeración.")
        if not (self.humedad_min <= datos["Humedad"] <= self.humedad_max):
            alertas.append("⚠️ ALERTA: Humedad fuera de rango. Ajustando vapor.")
        if datos["Torque"] >= self.torque_objetivo:
            alertas.append("✅ ÉXITO: Torque alcanzado. Masa lista para división.")
        return alertas
```

```python
# Inicializar sistema
sistema = SistemaControlPanIntegral()
historial = {"Temp": [], "Hum": [], "Torque": []}

try:
    for i in range(20): # Simular 20 ciclos de lectura
        datos = sistema.simular_lectura()
        alertas = sistema.verificar_alertas(datos)

        # Guardar historial
        historial["Temp"].append(datos["Temperatura"])
        historial["Hum"].append(datos["Humedad"])
        historial["Torque"].append(datos["Torque"])

        # Visualización
        clear_output(wait=True)
        fig, axs = plt.subplots(1, 3, figsize=(18, 5))

        axs[0].plot(historial["Temp"], color='red', marker='o')
        axs[0].set_title("Temperatura (°C)")
        axs[0].axhline(y=26, color='black', linestyle='--')

        axs[1].plot(historial["Hum"], color='blue', marker='o')
        axs[1].set_title("Humedad (%)")

        axs[2].plot(historial["Torque"], color='green', marker='o')
        axs[2].set_title("Torque de Amasado (Nm)")

        plt.show()

        # Imprimir estados y alertas en consola
        print(f"Lectura actual: {datos}")
        for a in alertas:
            print(a)

        time.sleep(1)

except KeyboardInterrupt:
    print("Monitoreo finalizado por el usuario.")
```
