```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from datetime import datetime, timedelta

# 1. GENERACIÓN DE DATASET SINTÉTICO (PANDAS & NUMPY)
def generate_industrial_dataset(records=500):
    np.random.seed(42)
    start_time = datetime.now()
    timestamps = [start_time + timedelta(minutes=i) for i in range(records)]

    # Lógica de señales: Base + Tendencia + Ruido
    vibration = 2.5 + np.random.normal(0, 0.6, records) # Base 2.5 mm/s
    temperature = 65 + 10 * np.sin(np.linspace(0, 4 * np.pi, records)) + np.random.normal(0, 1.2, records)
    pressure = 2200 + np.random.normal(0, 150, records) # Base 2200 psi

    # Inserción intencional de anomalías (5 picos por sensor)
    for _ in range(5):
        vibration[np.random.randint(0, records)] += 4.0
        temperature[np.random.randint(0, records)] += 25.0
        pressure[np.random.randint(0, records)] += 1000.0

    df = pd.DataFrame({
        'Timestamp': timestamps,
        'Vibracion_mms': vibration,
        'Temperatura_C': temperature,
        'Presion_psi': pressure
    })
    return df

df_industrial = generate_industrial_dataset()

# 2. ANÁLISIS ESTADÍSTICO (SEABORN)
def plot_statistical_analysis(df):
    sns.set_theme(style="whitegrid")
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle('Análisis Estadístico de Sensores Industriales', fontsize=16)

    # Distribuciones
    sns.histplot(df['Vibracion_mms'], kde=True, ax=axes[0, 0], color='skyblue').set_title('Distribución Vibración')
    sns.histplot(df['Temperatura_C'], kde=True, ax=axes[0, 1], color='salmon').set_title('Distribución Temperatura')
    sns.histplot(df['Presion_psi'], kde=True, ax=axes[1, 0], color='olive').set_title('Distribución Presión')

    # Heatmap de Correlación
    sns.heatmap(df.corr(), annot=True, cmap='coolwarm', ax=axes[1, 1])
    axes[1, 1].set_title('Matriz de Correlación')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# 3. DASHBOARD INTERACTIVO (PLOTLY)
def create_interactive_dashboard(df):
    fig = make_subplots(rows=3, cols=1, shared_xaxes=True,
                        vertical_spacing=0.08,
                        subplot_titles=("Nivel de Vibración", "Temperatura de Proceso", "Presión Hidráulica"))

    # Configuración de trazos
    fig.add_trace(go.Scatter(x=df['Timestamp'], y=df['Vibracion_mms'], name="Vib (mm/s)", line=dict(color='#00CC96')), row=1, col=1)
    fig.add_trace(go.Scatter(x=df['Timestamp'], y=df['Temperatura_C'], name="Temp (°C)", line=dict(color='#EF553B')), row=2, col=1)
    fig.add_trace(go.Scatter(x=df['Timestamp'], y=df['Presion_psi'], name="Pres (psi)", line=dict(color='#636EFA')), row=3, col=1)

    # Añadir Thresholds (Umbrales)
    fig.add_hline(y=5.0, line_dash="dash", line_color="red", row=1, col=1)
    fig.add_hline(y=85.0, line_dash="dash", line_color="red", row=2, col=1)
    fig.add_hline(y=3000, line_dash="dash", line_color="red", row=3, col=1)
    fig.add_hline(y=1500, line_dash="dash", line_color="red", row=3, col=1)

    fig.update_layout(height=800, title_text="Panel de Control Industrial IIoT", template="plotly_dark", showlegend=True)
    fig.show()

# 4. LÓGICA DE CONTROL Y ALERTAS
def run_security_audit(df):
    print("--- INICIANDO AUDITORÍA DE SEGURIDAD ---")
    alertas = []

    for i, row in df.iterrows():
        if row['Vibracion_mms'] > 5.0:
            alertas.append(f"ALERTA VIBRACIÓN: {row['Vibracion_mms']:.2f} mm/s en el índice {i}")
        if row['Temperatura_C'] > 85.0:
            alertas.append(f"ALERTA TEMPERATURA: {row['Temperatura_C']:.2f} °C en el índice {i}")
        if row['Presion_psi'] > 3000 or row['Presion_psi'] < 1500:
            alertas.append(f"ALERTA PRESIÓN: {row['Presion_psi']:.2f} psi en el índice {i}")

    if not alertas:
        print("No se detectaron anomalías.")
    else:
        for a in alertas:
            print(a)
    print(f"--- TOTAL DE ALERTAS DETECTADAS: {len(alertas)} ---")

# EJECUCIÓN DEL SISTEMA
if __name__ == "__main__":
    plot_statistical_analysis(df_industrial)
    create_interactive_dashboard(df_industrial)
    run_security_audit(df_industrial)
```
